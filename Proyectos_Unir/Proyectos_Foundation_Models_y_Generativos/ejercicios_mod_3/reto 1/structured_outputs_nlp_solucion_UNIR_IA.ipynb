```python
# Importar las librerías necesarias
from pydantic import BaseModel
import openai

# Configurar la clave de API de OpenAI
openai.api_key = "TU_API_KEY"

# Crear el esquema para la salida estructurada
class ProductDescription(BaseModel):
    product_name: str
    description: str
    seo_tags: list[str]

# Función para generar la descripción del producto
def generate_product_description(prompt):
    try:
        # Definir el esquema en formato JSON
        schema = {
            "type": "object",
            "properties": {
                "product_name": {"type": "string"},
                "description": {"type": "string"},
                "seo_tags": {"type": "array", "items": {"type": "string"}}
            },
            "required": ["product_name", "description", "seo_tags"]
        }

        # Solicitar la generación de salida estructurada
        completion = openai.ChatCompletion.create(
            model="gpt-4-0613",
            messages=[
                {"role": "system", "content": "Eres un modelo NLP especializado en generar descripciones de productos con optimización SEO."},
                {"role": "user", "content": prompt}
            ],
            functions=[
                {
                    "name": "generate_product_description",
                    "parameters": schema
                }
            ],
            function_call={"name": "generate_product_description"}  # Obligatorio para usar Structured Outputs
        )

        # Obtener la respuesta en formato JSON
        structured_output = completion.choices[0].message["function_call"]["arguments"]

        # Validar y procesar la salida usando Pydantic
        return ProductDescription.model_validate_json(structured_output)

    except Exception as e:
        print("Error al generar la descripción:", e)
        return None

# Probar la función con un producto de ejemplo
prompt = "Genera una descripción para un smartphone llamado 'TechMax Pro 2024' con las siguientes características: pantalla OLED, cámara de 108 MP, y batería de larga duración. Incluye etiquetas SEO relevantes."

response = generate_product_description(prompt)

# Mostrar la salida
if response:
    print("Descripción del producto:")
    print(f"Nombre: {response.product_name}")
    print(f"Descripción: {response.description}")
    print("Etiquetas SEO:", ", ".join(response.seo_tags))
```
