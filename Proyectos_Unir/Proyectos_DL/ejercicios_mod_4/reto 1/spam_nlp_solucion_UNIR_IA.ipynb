```python
# Importar las librerías necesarias
import pandas as pd
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, confusion_matrix

# 1. Cargar el conjunto de datos desde una ruta absoluta
df = pd.read_csv('spam.csv', encoding='latin-1')

# 2. Limpieza de los datos
# Mantener solo las columnas necesarias (etiqueta y mensaje)
df = df[['v1', 'v2']]
df.columns = ['label', 'message']  # Renombrar las columnas

# 3. Preprocesamiento del texto
# Definir una función para limpiar y preprocesar el texto
def preprocess_text(text):
    # Convertir a minúsculas
    text = text.lower()
    
    # Eliminar caracteres especiales y números
    text = re.sub(r'\W', ' ', text)  # Eliminar caracteres no alfanuméricos
    text = re.sub(r'\s+', ' ', text)  # Eliminar múltiples espacios en blanco
    text = text.strip()  # Eliminar espacios adicionales

    # Tokenización (en este caso, las palabras ya están divididas por espacio tras el preprocesamiento)
    tokens = text.split()

    # Eliminar stopwords manualmente (opcional) - se puede hacer una lista manual de stopwords o usar una librería
    stopwords = ['and', 'or', 'the', 'a', 'of', 'in', 'to', 'la', 'el', 'de']  # Ejemplo de stopwords
    filtered_tokens = [word for word in tokens if word not in stopwords]

    return ' '.join(filtered_tokens)

# Aplicar la función de preprocesamiento a los mensajes
df['message'] = df['message'].apply(preprocess_text)

# Visualizar los datos procesados
print(df.head())

# Representación de los mensajes con TF-IDF
vectorizer = TfidfVectorizer(max_features=5000)  # Limitar a 5000 características para optimizar el rendimiento
X = vectorizer.fit_transform(df['message'])

# 2. División del conjunto de datos en entrenamiento y validación
X_train, X_val, y_train, y_val = train_test_split(X, df['label'], test_size=0.2, random_state=42)

# Entrenar el modelo Naive Bayes
model = MultinomialNB()
model.fit(X_train, y_train)

# 2. Evaluación del rendimiento
y_pred = model.predict(X_val)

# Mostrar las métricas de evaluación
print("Reporte de Clasificación:\n", classification_report(y_val, y_pred))
print("Matriz de Confusión:\n", confusion_matrix(y_val, y_pred))

# Precisión del modelo
accuracy = model.score(X_val, y_val)
print(f'Precisión del modelo: {accuracy * 100:.2f}%')

# Evaluar el modelo con precisión, recall, F1-score, y matriz de confusión
print("Reporte de Clasificación:\n", classification_report(y_val, y_pred))
print("Matriz de Confusión:\n", confusion_matrix(y_val, y_pred))


```
