# 🛒 Análisis de Ventas con Pandas
**Herramientas:** Python · Pandas  
**Objetivo:** Explorar y manipular un DataFrame de ventas por producto y categoría, calculando totales, aplicando descuentos y filtrando datos.


## 1. Importar librerías

```python
import pandas as pd
```

## 2. Crear el DataFrame

Creamos el DataFrame a partir de una lista de diccionarios con datos de ventas por producto.

```python
ventas = [
    {"Producto": "Laptop", "Categoría": "Electrónica", "Unidades": 5, "Precio_unitario": 800},
    {"Producto": "Smartphone", "Categoría": "Electrónica", "Unidades": 10, "Precio_unitario": 600},
    {"Producto": "Impresora", "Categoría": "Oficina", "Unidades": 7, "Precio_unitario": 150},
    {"Producto": "Monitor", "Categoría": "Electrónica", "Unidades": 3, "Precio_unitario": 200},
    {"Producto": "Teclado", "Categoría": "Oficina", "Unidades": 15, "Precio_unitario": 25},
    {"Producto": "Smartphone", "Categoría": "Electrónica", "Unidades": 4, "Precio_unitario": 600},
    {"Producto": "Laptop", "Categoría": "Electrónica", "Unidades": 2, "Precio_unitario": 800},
    {"Producto": "Teclado", "Categoría": "Oficina", "Unidades": 10, "Precio_unitario": 25}
]

df = pd.DataFrame(ventas)
```

## 3. Exploración del DataFrame

Inspeccionamos las primeras y últimas filas, el resumen estadístico, el conteo por categoría y la estructura del DataFrame.

```python
# Primeras y últimas 5 filas
print("--- Primeras 5 filas ---")
print(df.head())
print("\n--- Últimas 5 filas ---")
print(df.tail())

# Resumen estadístico
print("\n--- Resumen estadístico ---")
print(df.describe())

# Productos por categoría
print("\n--- Productos por categoría ---")
print(df["Categoría"].value_counts())

# Estructura del DataFrame
print("\n--- Información del DataFrame ---")
df.info()
```

## 4. Columna Total

Calculamos el total de ventas por fila multiplicando `Unidades` por `Precio_unitario`.

```python
df["Total"] = df["Unidades"] * df["Precio_unitario"]
print("--- DataFrame con columna Total ---")
print(df)
```

## 5. Descuento para productos de Oficina

Aplicamos un descuento del 10% sobre el `Precio_unitario` de los productos de la categoría Oficina usando `loc`.

```python
# Convertimos a float para poder aplicar el descuento
df["Precio_unitario"] = df["Precio_unitario"].astype(float)

# Descuento del 10% solo para productos de Oficina
df.loc[df['Categoría'] == 'Oficina', 'Precio_unitario'] *= 0.90

print("--- DataFrame con descuento ---")
print(df)
```

## 6. Filtrar productos de Electrónica con más de 5 unidades

Filtramos el DataFrame con dos condiciones simultáneas: categoría `Electrónica` y más de 5 unidades vendidas.

```python
electronica_top = df[(df["Categoría"] == "Electrónica") & (df["Unidades"] > 5)]
print("--- Electrónica con más de 5 unidades ---")
print(electronica_top)
```
